```python
import tensorflow as tf

import numpy as np

import matplotlib.pyplot as plt


# 1. Create Training Data for y = 2x

# Standardized range of input features (x) and targets (y)

x_train = np.array([-3.0, -2.0, -1.0, 0.0, 1.0, 2.0, 3.0, 4.0], dtype=float)

y_train = np.array([-6.0, -4.0, -2.0, 0.0, 2.0, 4.0, 6.0, 8.0], dtype=float)


# 2. Build a Simple Dense Neural Network

# Single layer with 1 neuron expecting 1 input feature

model = tf.keras.Sequential([

tf.keras.layers.Dense(units=1, input_shape=[1])

])


# 3. Compile Model

# Stochastic Gradient Descent (SGD) optimizer with Mean Squared Error (MSE) loss

model.compile(optimizer='sgd', loss='mean_squared_error')


# 4. Train Model

print("Training the model...")

history = model.fit(x_train, y_train, epochs=500, verbose=0)

print("Training complete!")


# 5. Generate Test Predictions

x_test = np.linspace(-5, 5, 100)

y_actual = 2 * x_test

y_pred = model.predict(x_test)


# 6. Plot Predictions vs Actual

plt.figure(figsize=(8, 5))

plt.plot(x_test, y_actual, label='Actual Rule (y = 2x)', color='blue', linewidth=2)

plt.scatter(x_train, y_train, color='red', label='Training Data Points', zorder=5)

plt.plot(x_test, y_pred, label='TensorFlow Model Predictions', color='green', linestyle='--', linewidth=2)


plt.title('TensorFlow Model: Learning y = 2x')

plt.xlabel('x')

plt.ylabel('y')

plt.legend()

plt.grid(True)

plt.show()


# Print learned weights and bias

weight, bias = model.layers[0].get_weights()

print(f"Learned Weight (Slope): {weight[0][0]:.4f} (Expected: ~2.0)")

print(f"Learned Bias (Intercept): {bias[0]:.4f} (Expected: ~0.0)")
```
